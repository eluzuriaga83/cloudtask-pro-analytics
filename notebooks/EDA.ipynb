```python
subscriptions['churned']=(subscriptions['churned']=='Yes').astype(int)
```

```python
subscriptions['upgraded'] = (subscriptions['upgraded'] == 'Yes').astype(int)
```

```python
subscriptions['churn_date'] = pd.to_datetime(subscriptions['churn_date'], errors='coerce')
```

```python
subscriptions['signup_date'] = pd.to_datetime(subscriptions['signup_date'], errors='coerce')
```

```python
print(subscriptions.signup_date.max())
print(subscriptions.signup_date.min())
```

```python
subscriptions.info()
```

Validations

```python
print(f"\nDate range:{subscriptions.signup_date.min()} to {subscriptions.signup_date.max()}")
print(f"total customers: {subscriptions['customer_id'].nunique()}")
print(f"Total churn rate: {subscriptions['churned'].mean():.2%}")
print(f"Churned reasons: {subscriptions['churn_reason'].value_counts()}")

print(f"\nChurn by plan:")
for plan, rate in subscriptions.groupby('plan')['churned'].mean().items(): #Convert the result to a list of (plan, rate) pairs.
    print(f" {plan}: {rate:.2%}")
```

```python
# Business logic validation
assert subscriptions['monthly_revenue'].min() >= 0, "Revenue cannot be negative"
assert subscriptions['churned'].max() <= 1, "churned should be 0 or 1"
assert subscriptions['nps_score'].max() <= 10, "NPS should be 0-10"
assert subscriptions['feature_usage_pct'].max() <= 100, "Feature usage should be 0-100%"
```

```python
print(f" ✓ All business logic constraints met")
```
